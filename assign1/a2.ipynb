```python
import numpy as  np 
import pandas as pd 
import matplotlib.pyplot as plt 
```

```python
study_hours = np.random.uniform(1,10,100)
noise = np.random.normal(0,5,100)
test_score = 35 + 6 * study_hours + noise

df = pd.DataFrame({
    "sh" : study_hours,
    "ts" : test_score
})

print(df)

print(df.shape[0]) # total samples / rows
print(df.columns)  # column names
print(df.shape)    # rows,columns

print(df.describe()) # statistics

print(df.isnull().sum()) # mising values
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn import metrics

x = df[["sh"]]  # feature / independent
y = df["ts"]  # target / dependent

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)

model = LinearRegression()  # decide the model
model.fit(x_train, y_train) # train the model (que, ans to que)

slope = model.coef_[0]
inter = model.intercept_

y_pred = model.predict(x_test) # model predict on new que

mae = metrics.mean_absolute_error(y_test, y_pred) # (answer key, model predict)
mse = metrics.mean_squared_error(y_test, y_pred) 
r2 = metrics.r2_score(y_test, y_pred)

print(slope)
print(inter)
print(mae)
print(mse)
print(r2)
```

```python
new_df = pd.DataFrame({
    "Study_Hours" : x_test["sh"].values,
    "Actual data" : y_test.values,
    "Predicted data" : y_pred
})

new_df.to_csv("linearregressionpredictions.csv")
print(new_df)
```
