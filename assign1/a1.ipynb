```python
import numpy as np
import pandas as pd
```

```python
data = np.array([
    [10, -20, 30], 
    [40, -50, 60], 
    [70, 80, -90]])

print(data.mean(axis=1)) # each row mean
print(data.mean(axis=0)) # each column mean
print(data.std(axis=1))  # each row std
print(data.max(axis=0))  # each column max
print(data.min(axis=1))  # each row min
```

```python
A = np.array([ [2,-4], [-6,8]]) 
B = np.array([ [1,-3], [-5,-7]])

print(A+B) # sum
print(A-B) # sub
print(A*B) # dot product
print(A*3) # multiply 3 to each ele
print(A@B) # cross multiply
print(A.T) # transpose
print(np.linalg.det(A))  # determinatn
print(np.linalg.inv(A))  # inverse
```

```python
df = pd.DataFrame({
    "Products" : ["iPhone", "Laptop", "PC"],
    "price" : [150000,200000,250000],
    "Sales" : [5000,4000,1800],
    "Profit" : [80000,20000,50000]
})
print(df)  # complete DataFrame
print(df.columns)  # all column
print(df.shape)  # number of rows and columns
print(df.head(2))  # first 2 rows
# Alternative: print(df.iloc[:2])
print(df[["Sales", "Profit"]])  # Sales and Profit columns

print(df["Sales"].mean())
print(df.Sales.min())
print(df.Sales.max())

print(df[df["Profit"] > 4000])  # rows where Profit is greater than 4000
print(df.Products[df["Profit"] > 4000])  # only product names where Profit is greater than 4000
print(df.sort_values("Profit"))  # Sort the DataFrame by Profit in ascending order
```
