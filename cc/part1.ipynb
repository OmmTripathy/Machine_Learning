```python
np.arange(1, 11)          # 1,2,3...10  (end is NOT included, so 1 to 10)
np.arange(0, 10, 2)       # 0,2,4,6,8   (step of 2)
np.linspace(0, 1, 5)      # 5 evenly spaced numbers from 0 to 1
```

```python
np.random.rand(5)                   # 5 numbers between 0 and 1
np.random.uniform(10, 50, 10)       # 5 decimal numbers between 10 and 50
np.random.randint(10, 100, 10)      # 5 whole numbers from 10 to 99

np.random.normal(loc=50, scale=10, size=5)      # mean = 50, std = 10
```
