```python
import numpy as np 
import pandas as pd
from sklearn.model_selection import train_test_split 
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
np.random.seed(42)  # generate same random nos in every run

hours = np.random.uniform(1,10,100)
assignments = np.random.uniform(5,20,100)
marks = 30 + 6*hours + np.random.normal(0,5,100)

df = pd.DataFrame({
    "hours" : hours,
    "assignments" : assignments,
    "marks" : marks
})

q1 = df["marks"].quantile(0.25)
q2 = df["marks"].quantile(0.5)
q3 = df["marks"].quantile(0.75)

iqr = q3 - q2

low = q1 - 1.5*iqr
high = q3 + 1.5*iqr
outliers = df[(df["marks"] < low) | (df["marks"] > high)]
```

```python
x = df.drop("marks", axis=1)
y = df["marks"]

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)

model = LinearRegression()
model.fit(x_train, y_train)

model.coef_[0]
model.intercept_

y_train_pred = model.predict(x_train)
y_test_pred = model.predict(x_test)

mae_ = mean_absolute_error(y_train, y_train_pred)
mse_ = mean_squared_error(y_train, y_train_pred)
rmse_ = np.sqrt(mse_
)
r2_ = r2_score(y_train, y_train_pred)

mae = mean_absolute_error(y_test, y_test_pred)
mse = mean_squared_error(y_test, y_test_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_test_pred)
```

## --> why we predicted both training and testing data

The three situations
# 1. Underfitting (model is too simple) :

Train R² is low,
Test R² is low

# 2. Overfitting (model is too complex or memorizes) :

Train R² is very high,
Test R² is much lower

# 3. Best fit

Train R² is about 0.90
Test R² is about 0.88

```python
print("First Quartile (Q1):", q1)
print("Second Quartile (Q2):", q2)
print("Third Quartile (Q3):", q3)
print("Interquartile Range (IQR):", iqr)
print("Lower Bound:", low)
print("Upper Bound:", high)
print("Number of Outliers:", len(outliers))
print("Outliers:\n", outliers)

print("Training Features:\n", x_train.head())
print("Testing Features:\n", x_test.head())
print("Training Target:\n", y_train.head())
print("Testing Target:\n", y_test.head())

print("Model Coefficients:", model.coef_)
print("Model Intercept:", model.intercept_)

print("\nTraining Set Evaluation Metrics:")
print("MAE:", mae_)
print("MSE:", mse_)
print("RMSE:", rmse_)
print("R2 Score:", r2_)

print("\nTesting Set Evaluation Metrics:")
print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R2 Score:", r2)
```
