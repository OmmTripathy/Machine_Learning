```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split 
from sklearn.linear_model import LinearRegression 
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error, r2_score
```

```python
old = pd.read_csv("Student_Performance_Prediction.csv")
df = pd.read_csv("modified_Student_Performance.csv")

df["Assignments_Completed"] = old["Assignments_Completed"]

df["Academic_Activity"] = df["Practice_Tests"] + df["Assignments_Completed"]
```

```python
cols = ["Academic_Activity", "Practice_Tests", "Assignments_Completed"]
corr = df[cols].corrwith(df["Final_Score"])
print(corr)
```

```python
df.to_csv("feature_engineered_Student_Performance.csv", index=False)
```

```python
x = df.drop(columns = ["Final_Score"])
y = df["Final_Score"]
e = []

for d in [2,3]:
    poly_F = PolynomialFeatures(degree = d, include_bias = False)
    x_poly = poly_F.fit_transform(x)

    x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 42)

    model = LinearRegression()
    model.fit(x_train, y_train)

    print(d)
    y_pred_tr = model.predict(x_train)

    mse_tr = mean_squared_error(y_train, y_pred_tr)
    R2_tr = r2_score(y_train, y_pred_tr)

    print(mse_tr)
    print(R2_tr)

    y_pred = model.predict(x_test)

    mse = mean_squared_error(y_test, y_pred)
    R2 = r2_score(y_test, y_pred)

    print(mse)
    print(R2)

    e.append([d, mse_tr, mse, R2_tr, R2])
```

```python
results = pd.DataFrame(e, columns=["Degree", "Train MSE", "Test MSE", "Train R2", "Test R2"])
print(results.round(4))
```

```python
print("Lowest training MSE: degree", results.loc[results["Train MSE"].idxmin(), "Degree"])
print("Lowest testing MSE : degree", results.loc[results["Test MSE"].idxmin(), "Degree"])
print("Highest testing R2 : degree", results.loc[results["Test R2"].idxmax(), "Degree"])
```
